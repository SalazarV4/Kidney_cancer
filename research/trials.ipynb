```python
import torchvision.io as io
from torchvision.transforms import v2
import torch
from pathlib import Path
import numpy as np
```

```python
model = torch.load(f=Path("artifacts/model_training/model.pth"), weights_only=False)
```

```python
image = io.read_image(path=r"artifacts\data_ingestion\Kidney_dataset\Train\Tumor\Tumor- (458).jpg")
```

```python
transforms = v2.Compose([
            v2.ToImage(),
            v2.Resize(size=[224,224]),
            v2.ToDtype(dtype=torch.float32,scale=True)])
```

```python
model.eval()
with torch.inference_mode():
    x = transforms(image.unsqueeze(dim=0))
    y_pred = model(x)
    result = np.argmax(y_pred.detach().numpy(),axis=1)

if result[0] == 1:
    return [{'image':'Tumor'}]
else:
    return [{'image':'Normal'}]
```
